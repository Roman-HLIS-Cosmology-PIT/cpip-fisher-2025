```python
def explore_space(space):
    FisherMeta.POSTERIOR = False
    case_ML = FisherCase(FisherMeta(space, to_sigma8=True))
    case_ML.compute_values_and_errors(store_mats=True)

    FisherMeta.POSTERIOR = True
    case_MAP = FisherCase(FisherMeta(space, to_sigma8=True))
    case_MAP.compute_values_and_errors(store_mats=True)

    for params in [["omegam", "sigma8", "ns", "omegab", "h0", "roman_A1_1", "roman_A1_2"],
                   ["roman_B1_"+str(i+1) for i in range(8)],
                   ["roman_DZ_S"+str(i+1) for i in range(8)],
                   ["roman_M"+str(i+1) for i in range(8)],
                   ["omegam", "sigma8", "ns", "omegab", "h0", "roman_B1_3", "roman_B1_6",
                    "roman_DZ_S3", "roman_DZ_S6", "roman_M3", "roman_M6", "roman_A1_1", "roman_A1_2"]]:
        make_corner_plots(case_ML, case_MAP, params)

# explore_space("real")
# explore_space("fourier")
```
